```python
from dotenv import load_dotenv
import os

load_dotenv()
print("OpenAI key loaded:", bool(os.getenv("OPENAI_API_KEY")))
```

```python
from phi.agent import Agent
from phi.tools import tool
from fpdf import FPDF
from pathlib import Path

@tool
def paragraph_to_pdf(paragraph: str, city_name: str = "City") -> str:
    """
    Converts a paragraph into a PDF and returns the full file path.
    """
    # Make city name safe for filename
    safe_city_name = city_name.replace(" ", "_")

    # Output folder
    output_dir = Path("weather_reports")
    output_dir.mkdir(exist_ok=True)

    pdf_file_path = output_dir / f"{safe_city_name}_Weather_Report.pdf"

    # Create PDF
    pdf = FPDF()
    pdf.add_page()
    pdf.set_font("Arial", size=12)
    pdf.multi_cell(0, 10, paragraph)
    pdf.output(str(pdf_file_path))

    # RETURN ONLY the FILE PATH, NOT A MESSAGE
    return str(pdf_file_path)

# Instantiate the PDF agent
pdf_agent = Agent(
    name="PDF Agent",
    tools=[paragraph_to_pdf],
    instructions=[
        "Receive a paragraph string and a city name, convert it to a PDF file.",
        "Return ONLY the file path of the generated PDF."
    ]
)
```
