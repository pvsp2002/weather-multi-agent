```python
from phi.agent import Agent, Tool
```

```python
from phi.tools import tool

@tool
def weather_to_paragraph(weather_str: str) -> str:
    lines = weather_str.split('\n')
    city = temp = condition = humidity = coords = ""

    for line in lines:
        if line.startswith('- **City**:'):
            city = line.split(':', 1)[1].strip()
        elif line.startswith('- **Temperature**:'):
            temp = line.split(':', 1)[1].strip()
        elif line.startswith('- **Condition**:'):
            condition = line.split(':', 1)[1].strip()
        elif line.startswith('- **Humidity**:'):
            humidity = line.split(':', 1)[1].strip()
        elif line.startswith('- **Coordinates**:'):
            coords = line.split(':', 1)[1].strip()

    paragraph = (
        f"The current weather in {city} is {condition} "
        f"with a temperature of {temp} and humidity of {humidity}."
    )

    if coords:
        paragraph += f" The coordinates are {coords}."

    return paragraph
```

```python
from phi.agent import Agent

content_agent = Agent(
    name="Content Agent",
    tools=[weather_to_paragraph],
    instructions=[
        "You will receive structured weather data.",
        "Use the tool to convert it into a single readable paragraph.",
        "Do not add new facts.",
        "Return only the paragraph."
    ]
)
```
